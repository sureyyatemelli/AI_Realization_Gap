```python
# ============================================================
# AI REALIZATION GAP – EU27
# STEP 01: PROJECT INITIALIZATION
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
from datetime import datetime
import pandas as pd
import os

# ------------------------------------------------------------
# 1. PROJECT ROOT
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/AI_Realization_Gap_EU27"
)

folders = [
    "00_admin",
    "01_raw_data",
    "02_intermediate_data",
    "03_processed_data",
    "04_code",
    "05_tables",
    "06_figures",
    "07_models",
    "08_logs",
    "09_manuscript"
]

for folder in folders:
    (PROJECT_ROOT / folder).mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. EU-27 COUNTRY MASTER
# ------------------------------------------------------------

EU27 = {
    "AT": "Austria",
    "BE": "Belgium",
    "BG": "Bulgaria",
    "HR": "Croatia",
    "CY": "Cyprus",
    "CZ": "Czechia",
    "DK": "Denmark",
    "EE": "Estonia",
    "FI": "Finland",
    "FR": "France",
    "DE": "Germany",
    "EL": "Greece",
    "HU": "Hungary",
    "IE": "Ireland",
    "IT": "Italy",
    "LV": "Latvia",
    "LT": "Lithuania",
    "LU": "Luxembourg",
    "MT": "Malta",
    "NL": "Netherlands",
    "PL": "Poland",
    "PT": "Portugal",
    "RO": "Romania",
    "SK": "Slovakia",
    "SI": "Slovenia",
    "ES": "Spain",
    "SE": "Sweden"
}

country_master = pd.DataFrame(
    list(EU27.items()),
    columns=["geo", "country"]
)

country_master["iso3"] = [
    "AUT","BEL","BGR","HRV","CYP","CZE","DNK","EST","FIN",
    "FRA","DEU","GRC","HUN","IRL","ITA","LVA","LTU","LUX",
    "MLT","NLD","POL","PRT","ROU","SVK","SVN","ESP","SWE"
]

# ------------------------------------------------------------
# 3. ANALYSIS YEARS
# ------------------------------------------------------------

YEARS = list(range(2021, 2026))

country_year_master = (
    country_master.assign(key=1)
    .merge(
        pd.DataFrame({"year": YEARS, "key": 1}),
        on="key"
    )
    .drop(columns="key")
    .sort_values(["country", "year"])
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 4. SAVE MASTER FILES
# ------------------------------------------------------------

country_master.to_csv(
    PROJECT_ROOT / "00_admin" / "EU27_country_master.csv",
    index=False
)

country_year_master.to_csv(
    PROJECT_ROOT / "02_intermediate_data" /
    "Step_01_EU27_country_year_master_2021_2025.csv",
    index=False
)

# ------------------------------------------------------------
# 5. PROJECT METADATA
# ------------------------------------------------------------

metadata = pd.DataFrame({
    "item": [
        "project",
        "geographical_scope",
        "number_of_countries",
        "target_period",
        "target_observations",
        "created"
    ],
    "value": [
        "AI Realization Gap",
        "EU-27",
        27,
        "2021-2025",
        27 * 5,
        datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    ]
})

metadata.to_csv(
    PROJECT_ROOT / "00_admin" / "project_metadata.csv",
    index=False
)

print("=" * 60)
print("AI REALIZATION GAP PROJECT INITIALIZED")
print("=" * 60)
print("Project root :", PROJECT_ROOT)
print("Countries    :", len(country_master))
print("Years        :", YEARS)
print("Potential N  :", len(country_year_master))
print("=" * 60)

display(country_year_master.head(10))
```
