```python
# STEP 02 — AI ADOPTION (EU-27) | GitHub-ready
# Run Step 01 first. Google Colab writes outputs to Google Drive.
from pathlib import Path
import json, itertools, zipfile, requests
import numpy as np
import pandas as pd
from IPython.display import display

try:
    from google.colab import drive
    drive.mount('/content/drive')
    PROJECT_ROOT = Path('/content/drive/MyDrive/AI_Realization_Gap_EU27')
except ImportError:
    # Local runs: set PROJECT_ROOT to your matching Step 01 project folder.
    PROJECT_ROOT = Path.cwd()

ADMIN_DIR = PROJECT_ROOT / '00_admin'
RAW_DIR = PROJECT_ROOT / '01_raw_data'
INTERMEDIATE_DIR = PROJECT_ROOT / '02_intermediate_data'
PROCESSED_DIR = PROJECT_ROOT / '03_processed_data'
TABLE_DIR = PROJECT_ROOT / '05_tables'
LOG_DIR = PROJECT_ROOT / '08_logs'
for folder in [ADMIN_DIR, RAW_DIR, INTERMEDIATE_DIR, PROCESSED_DIR, TABLE_DIR, LOG_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

# True: use saved Eurostat JSON snapshot for exact historical reproduction.
# False: fetch current Eurostat data; revisions can change results.
USE_SAVED_SNAPSHOT = True
print('PROJECT_ROOT:', PROJECT_ROOT)
```

```python
country_master_path = ADMIN_DIR / 'eu27_country_master.csv'
if not country_master_path.exists():
    raise FileNotFoundError(f'Run Step 01 first: {country_master_path}')
country_master = pd.read_csv(country_master_path)
EU27 = country_master['geo'].tolist()
assert len(EU27) == len(set(EU27)) == 27, 'Country master must list 27 unique countries.'
print('EU-27 countries:', len(EU27))
display(country_master.head())
```

```python
AI_YEARS = [2021, 2023, 2024, 2025]
print('Observation years:', AI_YEARS)
```

```python
DATASET = 'isoc_eb_ai'
url = ('https://ec.europa.eu/eurostat/api/dissemination/'
       f'statistics/1.0/data/{DATASET}?lang=en&sinceTimePeriod=2021&untilTimePeriod=2025')
raw_json_path = RAW_DIR / 'Step_02_Eurostat_isoc_eb_ai_raw.json'
if USE_SAVED_SNAPSHOT and raw_json_path.exists():
    with raw_json_path.open('r', encoding='utf-8') as f:
        data = json.load(f)
    print('Using archived Eurostat JSON:', raw_json_path)
else:
    response = requests.get(url, timeout=120)
    response.raise_for_status()
    data = response.json()
    with raw_json_path.open('w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False)
    print('Downloaded Eurostat dataset and saved snapshot:', raw_json_path)
print('Dataset:', DATASET, '| Updated:', data.get('updated', 'not supplied'))
```

```python
# The raw JSON was saved or reused by the previous cell.
assert raw_json_path.exists(), 'Raw snapshot not found'
print('Raw data:', raw_json_path)
```

```python
# ============================================================
# 7. INSPECT EUROSTAT DIMENSIONS
# ============================================================

print("Dimension order:")
print(data["id"])

print("\nDimension sizes:")
print(data["size"])

for dim in data["id"]:

    print("\n" + "=" * 60)
    print("DIMENSION:", dim)

    dim_info = data["dimension"][dim]

    print("Label:", dim_info.get("label"))

    categories = (
        dim_info
        .get("category", {})
        .get("label", {})
    )

    # Show first categories only
    for code, label in list(categories.items())[:20]:
        print(f"{code}: {label}")
```

```python
def ordered_codes(dataset, dim):
    index = dataset['dimension'][dim]['category']['index']
    if isinstance(index, dict):
        return [code for code, position in sorted(index.items(), key=lambda x: x[1])]
    if isinstance(index, list):
        return index
    raise TypeError(f'Unexpected Eurostat category index type: {dim}')

dimension_ids = data['id']
dimension_codes = {dim: ordered_codes(data, dim) for dim in dimension_ids}
assert all(len(dimension_codes[d]) == data['size'][i] for i, d in enumerate(dimension_ids))
print('Eurostat dimensions:', {d: len(dimension_codes[d]) for d in dimension_ids})
```

```python
# Decode only the populated JSON-stat cells (avoids an enormous Cartesian grid).
# JSON-stat uses row-major indexing with the final dimension changing fastest.
shape = [len(dimension_codes[d]) for d in dimension_ids]
values_raw = data.get('value', {})
records = []
for key, value in values_raw.items():
    remainder = int(key)
    indexes = [0] * len(shape)
    for j in range(len(shape) - 1, -1, -1):
        indexes[j] = remainder % shape[j]
        remainder //= shape[j]
    if remainder:
        raise ValueError(f'JSON-stat index out of range: {key}')
    records.append({**{d: dimension_codes[d][idx] for d, idx in zip(dimension_ids, indexes)}, 'value': value})
df_long = pd.DataFrame(records)
print('Populated Eurostat cells:', len(df_long))
display(df_long.head())
```

```python
# ============================================================
# 10. CHECK RELEVANT EUROSTAT CATEGORIES
# ============================================================

for variable in [
    "indic_is",
    "size_emp",
    "unit",
    "nace_r2"
]:

    if variable in df_long.columns:

        print("\n", "=" * 60)
        print(variable)

        print(
            sorted(
                df_long[variable]
                .dropna()
                .unique()
                .tolist()
            )
        )
```

```python
# ============================================================
# 11. INSPECT NACE CATEGORIES
# ============================================================

nace_labels = (
    data["dimension"]["nace_r2"]
    ["category"]
    .get("label", {})
)

nace_table = pd.DataFrame({
    "nace_r2": list(nace_labels.keys()),
    "label": list(nace_labels.values())
})

display(nace_table)
```

```python
# The specific broad NACE category is selected by its unique classification code,
# not by potentially ambiguous text matching.
print(nace_table[nace_table['label'].str.contains('All activities|Total', case=False, na=False)])
```

```python
TOTAL_NACE = 'C10-S951_X_K'
if TOTAL_NACE not in dimension_codes['nace_r2']:
    raise ValueError(f'Required NACE category absent: {TOTAL_NACE}')
print('NACE selection:', TOTAL_NACE,
      data['dimension']['nace_r2']['category']['label'].get(TOTAL_NACE))
```

```python
# Explicitly fixed category: prevents stale TOTAL_NACE state in Colab.
assert TOTAL_NACE == 'C10-S951_X_K'
```

```python
required_codes = {'indic_is': 'E_AI_TANY', 'size_emp': 'GE10',
                  'unit': 'PC_ENT', 'nace_r2': TOTAL_NACE}
for dim, code in required_codes.items():
    assert code in dimension_codes[dim], f'Missing Eurostat category {dim}={code}'
ai = df_long.loc[
    df_long['geo'].isin(EU27)
    & df_long['time'].astype(str).isin([str(y) for y in AI_YEARS])
    & (df_long['indic_is'] == required_codes['indic_is'])
    & (df_long['size_emp'] == required_codes['size_emp'])
    & (df_long['unit'] == required_codes['unit'])
    & (df_long['nace_r2'] == required_codes['nace_r2'])
].copy()
ai['year'] = ai['time'].astype(int)
ai['ai_adoption_pct'] = pd.to_numeric(ai['value'], errors='coerce')
ai = ai[['geo', 'year', 'ai_adoption_pct']].sort_values(['geo','year']).reset_index(drop=True)
assert not ai.duplicated(['geo', 'year']).any(), 'Duplicated Eurostat country-year records'
print('Filtered observations:', len(ai))
display(ai.head())
```

```python
skeleton_path = INTERMEDIATE_DIR / 'eu27_ai_year_panel_skeleton.csv'
if not skeleton_path.exists():
    raise FileNotFoundError(f'Step 01 panel skeleton not found: {skeleton_path}')
panel = pd.read_csv(skeleton_path)
assert set(panel['year'].unique()) == set(AI_YEARS)
assert len(panel) == 108 and not panel.duplicated(['geo','year']).any()
ai_panel = panel.merge(ai, on=['geo', 'year'], how='left', validate='one_to_one')
print('Merged panel:', ai_panel.shape)
display(ai_panel.head())
```

```python
# ============================================================
# 15. MISSING-VALUE CHECK
# ============================================================

missing_ai = ai_panel[
    ai_panel["ai_adoption_pct"].isna()
].copy()

print(
    "Missing AI adoption observations:",
    len(missing_ai)
)

if len(missing_ai) > 0:

    display(missing_ai)

else:

    print(
        "No missing AI adoption observations."
    )
```

```python
# ============================================================
# 16. YEARLY SUMMARY
# ============================================================

year_summary = (
    ai_panel
    .groupby("year")
    ["ai_adoption_pct"]
    .agg(
        N="count",
        mean="mean",
        median="median",
        std="std",
        min="min",
        max="max"
    )
    .reset_index()
)

display(year_summary)
```

```python
# ============================================================
# 17. CREATE 0-1 AI ADOPTION RATE
# ============================================================

ai_panel["ai_adoption"] = (
    ai_panel["ai_adoption_pct"] / 100
)

display(
    ai_panel[
        [
            "country",
            "year",
            "ai_adoption_pct",
            "ai_adoption"
        ]
    ].head(12)
)
```

```python
# ============================================================
# 18. DUPLICATE CHECK
# ============================================================

duplicates = ai_panel.duplicated(
    subset=["iso3", "year"],
    keep=False
)

print(
    "Duplicated country-year observations:",
    duplicates.sum()
)

if duplicates.any():
    display(ai_panel[duplicates])
```

```python
# ============================================================
# 19. SAVE FINAL AI ADOPTION PANEL
# ============================================================

output_path = (
    PROCESSED_DIR /
    "Step_02_AI_adoption_EU27_2021_2023_2024_2025.csv"
)

ai_panel.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("Saved:")
print(output_path)
```

```python
# ============================================================
# 20. SAVE DESCRIPTIVE TABLE
# ============================================================

summary_path = (
    TABLE_DIR /
    "Step_02_AI_adoption_year_summary.csv"
)

year_summary.to_csv(
    summary_path,
    index=False,
    encoding="utf-8-sig"
)

print("Saved:")
print(summary_path)
```

```python
# ============================================================
# 21. QUALITY CONTROL
# ============================================================

qc_checks = {

    "Panel contains 108 country-year rows":
        len(ai_panel) == 108,

    "EU27 contains 27 countries":
        ai_panel["iso3"].nunique() == 27,

    "Four AI observation years":
        ai_panel["year"].nunique() == 4,

    "Correct years":
        set(ai_panel["year"].unique())
        == set(AI_YEARS),

    "No duplicate country-year":
        ai_panel.duplicated(
            ["iso3", "year"]
        ).sum() == 0,

    "AI adoption percentage >= 0":
        (
            ai_panel["ai_adoption_pct"]
            .dropna() >= 0
        ).all(),

    "AI adoption percentage <= 100":
        (
            ai_panel["ai_adoption_pct"]
            .dropna() <= 100
        ).all(),

    "AI adoption rate >= 0":
        (
            ai_panel["ai_adoption"]
            .dropna() >= 0
        ).all(),

    "AI adoption rate <= 1":
        (
            ai_panel["ai_adoption"]
            .dropna() <= 1
        ).all()
}

qc = pd.DataFrame({
    "check": qc_checks.keys(),
    "passed": qc_checks.values()
})

display(qc)
if not qc['passed'].all():
    raise AssertionError('Step 02 QC failed: ' + ', '.join(qc.loc[~qc['passed'], 'check']))
```

```python
# ============================================================
# 22. SAVE QC REPORT
# ============================================================

qc_path = (
    LOG_DIR /
    "Step_02_QC.csv"
)

qc.to_csv(
    qc_path,
    index=False,
    encoding="utf-8-sig"
)

print("QC saved:")
print(qc_path)
```

```python
# ============================================================
# 23. STEP 02 FINAL REPORT
# ============================================================

print("=" * 68)
print("STEP 02 — AI ADOPTION COMPLETED")
print("=" * 68)

print(
    "Countries       :",
    ai_panel["iso3"].nunique()
)

print(
    "Years           :",
    sorted(ai_panel["year"].unique())
)

print(
    "Panel rows      :",
    len(ai_panel)
)

print(
    "Observed AI     :",
    ai_panel["ai_adoption_pct"].notna().sum()
)

print(
    "Missing AI      :",
    ai_panel["ai_adoption_pct"].isna().sum()
)

print("\nOutput:")
print(output_path)

print("=" * 68)
```
