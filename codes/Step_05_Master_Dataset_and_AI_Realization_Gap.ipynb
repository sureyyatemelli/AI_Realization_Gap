```python
# ============================================================
# AI REALIZATION GAP — EU27
# STEP 05 — MASTER DATASET & AI REALIZATION GAP
# ============================================================

from google.colab import drive
from pathlib import Path

import pandas as pd
import numpy as np
import json
from datetime import datetime

drive.mount("/content/drive")

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/AI_Realization_Gap_EU27"
)

ADMIN_DIR = PROJECT_ROOT / "00_admin"
INTERMEDIATE_DIR = PROJECT_ROOT / "02_intermediate_data"
PROCESSED_DIR = PROJECT_ROOT / "03_processed_data"
TABLE_DIR = PROJECT_ROOT / "05_tables"
LOG_DIR = PROJECT_ROOT / "08_logs"

AI_YEARS = [2021, 2023, 2024, 2025]

print("Project root:", PROJECT_ROOT)
print("Analysis years:", AI_YEARS)
```

```python
# ============================================================
# LOAD STEP 02 — AI ADOPTION
# ============================================================

step02_path = (
    PROCESSED_DIR /
    "Step_02_AI_adoption_EU27_2021_2023_2024_2025.csv"
)

adoption = pd.read_csv(step02_path)

print("Step 02:", adoption.shape)
display(adoption.head())
```

```python
# ============================================================
# LOAD STEP 03 — AI POTENTIAL
# ============================================================

step03_path = (
    PROCESSED_DIR /
    "Step_03_AI_potential_EU27_2021_2023_2024_2025.csv"
)

potential = pd.read_csv(step03_path)

print("Step 03:", potential.shape)
display(potential.head())
```

```python
# ============================================================
# LOAD STEP 04 — COMPLEMENTARY ENABLERS
# ============================================================

step04_path = (
    PROCESSED_DIR /
    "Step_04_Complementary_Enablers_EU27.csv"
)

enablers = pd.read_csv(step04_path)

print("Step 04:", enablers.shape)
display(enablers.head())
```

```python
# ============================================================
# CHECK REQUIRED VARIABLES
# ============================================================

required_adoption = [
    "geo",
    "country",
    "iso3",
    "year",
    "ai_adoption_pct",
    "ai_adoption"
]

required_potential = [
    "geo",
    "country",
    "iso3",
    "year",
    "ict_specialists_pct",
    "hrst_pct",
    "hightech_emp_pct",
    "ai_potential_pca_z",
    "ai_potential",
    "ai_potential_equal_z",
    "ai_potential_equal"
]

required_enablers = [
    "geo",
    "country",
    "iso3",
    "year",
    "vhcn_pct",
    "gdp_pc_pps",
    "ln_gdp_pc_pps"
]

for name, df, required in [
    ("Step 02", adoption, required_adoption),
    ("Step 03", potential, required_potential),
    ("Step 04", enablers, required_enablers)
]:

    missing_cols = [
        col for col in required
        if col not in df.columns
    ]

    print(name, "missing columns:", missing_cols)
```

```python
# ============================================================
# MERGE ADOPTION + POTENTIAL
# ============================================================

master = adoption[
    required_adoption
].merge(

    potential[
        required_potential
    ],

    on=[
        "geo",
        "country",
        "iso3",
        "year"
    ],

    how="inner",
    validate="one_to_one"
)

print("After Step 02 + Step 03:", master.shape)
```

```python
# ============================================================
# ADD COMPLEMENTARY ENABLERS
# ============================================================

master = master.merge(

    enablers[
        required_enablers
    ],

    on=[
        "geo",
        "country",
        "iso3",
        "year"
    ],

    how="inner",
    validate="one_to_one"
)

master = (
    master
    .sort_values(
        ["iso3", "year"]
    )
    .reset_index(drop=True)
)

print("Master panel shape:", master.shape)

display(master.head(15))
```

```python
# ============================================================
# MASTER DATASET BASIC QC
# ============================================================

print("Rows:", len(master))
print("Countries:", master["iso3"].nunique())
print("Years:", sorted(master["year"].unique()))

print(
    "Duplicates:",
    master.duplicated(
        ["iso3", "year"]
    ).sum()
)

print("\nMissing values:")

display(
    master
    .isna()
    .sum()
    .to_frame("missing_n")
)
```

```python
# ============================================================
# MIN-MAX NORMALIZE AI ADOPTION
# ACROSS ALL 108 COUNTRY-YEAR OBSERVATIONS
# ============================================================

adopt_min = master[
    "ai_adoption_pct"
].min()

adopt_max = master[
    "ai_adoption_pct"
].max()

master["ai_adoption_norm"] = (
    master["ai_adoption_pct"]
    - adopt_min
) / (
    adopt_max
    - adopt_min
)

print("Adoption original range:")
print(
    master["ai_adoption_pct"]
    .min(),
    master["ai_adoption_pct"]
    .max()
)

print("\nNormalized range:")
print(
    master["ai_adoption_norm"]
    .min(),
    master["ai_adoption_norm"]
    .max()
)
```

```python
# ============================================================
# MAIN AI REALIZATION GAP
# ============================================================

master["ai_realization_gap"] = (
    master["ai_potential"]
    -
    master["ai_adoption_norm"]
)

display(
    master[
        [
            "country",
            "year",
            "ai_potential",
            "ai_adoption_pct",
            "ai_adoption_norm",
            "ai_realization_gap"
        ]
    ].head(20)
)
```

```python
# ============================================================
# DESCRIPTIVE GAP CLASSIFICATION
# ============================================================

master["gap_direction"] = np.select(

    [
        master["ai_realization_gap"] > 0,
        master["ai_realization_gap"] < 0
    ],

    [
        "Potential exceeds adoption",
        "Adoption exceeds potential"
    ],

    default="Balanced"
)

display(
    master[
        [
            "country",
            "year",
            "ai_realization_gap",
            "gap_direction"
        ]
    ].head(20)
)
```

```python
# ============================================================
# STANDARDIZED ADOPTION
# ============================================================

adopt_mean = master[
    "ai_adoption_pct"
].mean()

adopt_std = master[
    "ai_adoption_pct"
].std(ddof=0)

master["z_ai_adoption"] = (
    master["ai_adoption_pct"]
    - adopt_mean
) / adopt_std
```

```python
# ============================================================
# Z-SCORE REALIZATION GAP — ROBUSTNESS
# ============================================================

master["ai_realization_gap_z"] = (
    master["ai_potential_pca_z"]
    -
    master["z_ai_adoption"]
)

display(
    master[
        [
            "country",
            "year",
            "ai_realization_gap",
            "ai_realization_gap_z"
        ]
    ].head(20)
)
```

```python
# ============================================================
# EQUAL-WEIGHT GAP — ROBUSTNESS
# ============================================================

master["ai_realization_gap_equal"] = (
    master["ai_potential_equal"]
    -
    master["ai_adoption_norm"]
)

display(
    master[
        [
            "country",
            "year",
            "ai_realization_gap",
            "ai_realization_gap_equal"
        ]
    ].head()
)
```

```python
# ============================================================
# ROBUSTNESS CORRELATION
# ============================================================

gap_corr = master[
    [
        "ai_realization_gap",
        "ai_realization_gap_z",
        "ai_realization_gap_equal"
    ]
].corr()

display(gap_corr)
```

```python
# ============================================================
# CORE DESCRIPTIVE STATISTICS
# ============================================================

core_vars = [
    "ai_adoption_pct",
    "ai_potential",
    "ai_realization_gap",
    "vhcn_pct",
    "gdp_pc_pps",
    "ln_gdp_pc_pps"
]

descriptive = (
    master[core_vars]
    .describe()
    .T
)

display(descriptive)
```

```python
# ============================================================
# YEARLY GAP SUMMARY
# ============================================================

gap_by_year = (
    master
    .groupby("year")
    ["ai_realization_gap"]
    .agg(
        N="count",
        mean="mean",
        median="median",
        std="std",
        min="min",
        max="max"
    )
    .reset_index()
)

display(gap_by_year)
```

```python
# ============================================================
# COUNTRY AVERAGE GAP
# ============================================================

gap_by_country = (
    master
    .groupby(
        ["iso3", "country"]
    )
    ["ai_realization_gap"]
    .agg(
        mean_gap="mean",
        median_gap="median",
        min_gap="min",
        max_gap="max"
    )
    .reset_index()
    .sort_values(
        "mean_gap",
        ascending=False
    )
)

display(gap_by_country)
```

```python
# ============================================================
# CHANGE IN REALIZATION GAP: 2021 → 2025
# ============================================================

gap_wide = (
    master[
        [
            "iso3",
            "country",
            "year",
            "ai_realization_gap"
        ]
    ]
    .pivot(
        index=["iso3", "country"],
        columns="year",
        values="ai_realization_gap"
    )
    .reset_index()
)

gap_wide["gap_change_2021_2025"] = (
    gap_wide[2025]
    -
    gap_wide[2021]
)

display(
    gap_wide.sort_values(
        "gap_change_2021_2025"
    )
)
```

```python
# ============================================================
# ADD CLOUD AS AUXILIARY VARIABLE
# ============================================================

cloud_path = (
    INTERMEDIATE_DIR /
    "Step_04_Cloud_auxiliary_EU27.csv"
)

cloud_aux = pd.read_csv(
    cloud_path
)

master = master.merge(

    cloud_aux[
        [
            "geo",
            "year",
            "cloud_pct"
        ]
    ],

    on=[
        "geo",
        "year"
    ],

    how="left",
    validate="one_to_one"
)

print(
    "Cloud observed:",
    master["cloud_pct"].notna().sum()
)

print(
    "Cloud missing:",
    master["cloud_pct"].isna().sum()
)
```

```python
# ============================================================
# FINAL COLUMN ORDER
# ============================================================

final_cols = [

    # identifiers
    "geo",
    "country",
    "iso3",
    "year",

    # realized AI
    "ai_adoption_pct",
    "ai_adoption",
    "ai_adoption_norm",
    "z_ai_adoption",

    # AI potential components
    "ict_specialists_pct",
    "hrst_pct",
    "hightech_emp_pct",

    # AI potential
    "ai_potential_pca_z",
    "ai_potential",
    "ai_potential_equal_z",
    "ai_potential_equal",

    # main outcome
    "ai_realization_gap",

    # robustness outcomes
    "ai_realization_gap_z",
    "ai_realization_gap_equal",

    # complementary enablers
    "vhcn_pct",
    "gdp_pc_pps",
    "ln_gdp_pc_pps",

    # auxiliary
    "cloud_pct"
]

master_final = (
    master[final_cols]
    .sort_values(
        ["iso3", "year"]
    )
    .reset_index(drop=True)
)

print(master_final.shape)

display(master_final.head(20))
```

```python
# ============================================================
# STEP 05 FINAL QC
# ============================================================

core_complete_vars = [
    "ai_adoption_pct",
    "ai_potential",
    "ai_realization_gap",
    "vhcn_pct",
    "gdp_pc_pps",
    "ln_gdp_pc_pps"
]

qc_checks = {

    "Panel N = 108":
        len(master_final) == 108,

    "Countries = 27":
        master_final[
            "iso3"
        ].nunique() == 27,

    "Years = 4":
        master_final[
            "year"
        ].nunique() == 4,

    "Correct years":
        set(
            master_final["year"]
        ) == {2021, 2023, 2024, 2025},

    "No duplicate country-year":
        master_final.duplicated(
            ["iso3", "year"]
        ).sum() == 0,

    "Core variables complete":
        master_final[
            core_complete_vars
        ].isna().sum().sum() == 0,

    "AI potential in [0,1]":
        (
            master_final[
                "ai_potential"
            ].between(0, 1)
        ).all(),

    "Normalized adoption in [0,1]":
        (
            master_final[
                "ai_adoption_norm"
            ].between(0, 1)
        ).all(),

    "Gap finite":
        np.isfinite(
            master_final[
                "ai_realization_gap"
            ]
        ).all()
}

qc = pd.DataFrame({
    "check": qc_checks.keys(),
    "passed": qc_checks.values()
})

display(qc)
```

```python
# ============================================================
# SAVE MASTER DATASET
# ============================================================

master_path = (
    PROCESSED_DIR /
    "Step_05_Master_AI_Realization_Gap_EU27.csv"
)

master_final.to_csv(
    master_path,
    index=False,
    encoding="utf-8-sig"
)

print("Saved:")
print(master_path)
```

```python
# ============================================================
# SAVE TABLES
# ============================================================

descriptive.to_csv(
    TABLE_DIR /
    "Step_05_Descriptive_Statistics.csv",
    encoding="utf-8-sig"
)

gap_by_year.to_csv(
    TABLE_DIR /
    "Step_05_Gap_by_Year.csv",
    index=False,
    encoding="utf-8-sig"
)

gap_by_country.to_csv(
    TABLE_DIR /
    "Step_05_Gap_by_Country.csv",
    index=False,
    encoding="utf-8-sig"
)

gap_corr.to_csv(
    TABLE_DIR /
    "Step_05_Gap_Robustness_Correlations.csv",
    encoding="utf-8-sig"
)
```
